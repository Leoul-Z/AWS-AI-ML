```python
import pandas as pd
import numpy as np

# Load the raw dataset
url = "https://raw.githubusercontent.com/plotly/datasets/master/supermarket_Sales.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
# Display column names
print("Column names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
def clean_data(data):
    # Create a copy to protect the original dataset

    data = data.copy()

    # Remove extra spaces from column names
    data.columns = data.columns.str.strip()

    # Remove duplicate rows
    data = data.drop_duplicates()

    # Remove rows containing missing values
    data = data.dropna()

    return data


# Clean the dataset
df_clean = clean_data(df)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

df_clean.head()
```

```python
def process_data(data):
    # Convert the Total column into a NumPy array
    sales = data["Total"].to_numpy(dtype=float)

    # Calculate statistics using NumPy
    results = {
        "Total Sales": np.sum(sales),
        "Average Sales": np.mean(sales),
        "Maximum Sale": np.max(sales),
        "Minimum Sale": np.min(sales),
        "Standard Deviation": np.std(sales)
    }

    return results


# Process the cleaned dataset
results = process_data(df_clean)

print("Data Processing Results:")
for key, value in results.items():
    print(f"{key}: {value:.2f}")
```

```python
# Calculate total sales for each product line
sales_by_product = df_clean.groupby("Product line")["Total"].sum()

print("Total Sales by Product Line:")
print(sales_by_product.sort_values(ascending=False))
```

```python
# Save the cleaned dataset
output_file = "cleaned_supermarket_sales.csv"

df_clean.to_csv(output_file, index=False)

print("Cleaned dataset exported successfully:", output_file)
```
